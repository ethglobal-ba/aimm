## REST API

```python
import requests
AGENT_URL = "http://localhost:5001" # Docker compose
# GET /
response = requests.get(f"{AGENT_URL}/")
print("GET / response:", response.text)


# POST /updatePrice
response = requests.post(f"{AGENT_URL}/updatePrice", json={})

print("POST /updatePrice response:", response.text)
```
